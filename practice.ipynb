```python
from langchain_google_genai import ChatGoogleGenerativeAI,GoogleGenerativeAIEmbeddings
from langchain_core.vectorstores import VectorStore
from langchain_community.vectorstores import Chroma,FAISS
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain.chains import RetrievalQA
from dotenv import load_dotenv
import os
load_dotenv()
```

# Loading the Data 

```python
DATA_DIR = "data"
docs = []
for filename in os.listdir(DATA_DIR):
    if filename.endswith(".txt"):
        loader = TextLoader(os.path.join(DATA_DIR, filename))
        docs.extend(loader.load())
```

# Now we are going to split the data 

```python
TextSpliter=RecursiveCharacterTextSplitter(chunk_size=450,chunk_overlap=140)
splits=TextSpliter.split_documents(docs)
```

#### Now we are going to convert this split documents into embeddings/vector to store in the vectorDB for it for search and performance

```python
embedding=GoogleGenerativeAIEmbeddings(model="gemini-embedding-001")
```

```python
texts = [doc.page_content for doc in splits]
embeddings=embedding.embed_documents(texts)

```

```python
vector_store=FAISS.from_texts(texts,embedding=embedding)
```

```python
retriver=vector_store.as_retriever(kwargs={"k":1})
```

```python
retriver.invoke("what is the one thing that Raghav hate most")
```
